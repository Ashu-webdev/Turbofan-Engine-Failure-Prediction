```python
threshold = 5

to_remove = [col for col in df.columns if df[col].nunique() < threshold]

print("Columns to remove:", to_remove)

df = df.drop(columns=to_remove)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.ensemble import ExtraTreesRegressor


X = df.drop('RUL', axis=1)
y = df['RUL']


etr = ExtraTreesRegressor(n_estimators=100, random_state=42)
etr.fit(X, y)


importances = pd.Series(etr.feature_importances_, index=X.columns).sort_values(ascending=False)


plt.figure(figsize=(14,6))
bars = plt.bar(importances.index, importances.values, color='skyblue')


plt.xticks(rotation=45, ha='right', fontsize=12)
plt.yticks(fontsize=12)
plt.ylabel("Importance", fontsize=14)
plt.title("Feature Importance using Extra Trees Regressor", fontsize=16)

plt.tight_layout()  
plt.show()

```

```python
new_df=df[['Cycle', 'SensorMeasure11', 'SensorMeasure13', 'SensorMeasure4',
       'SensorMeasure14', 'SensorMeasure15', 'SensorMeasure9',
       'SensorMeasure8', 'SensorMeasure3', 'SensorMeasure17',
       'SensorMeasure12', 'SensorMeasure7', 'SensorMeasure2', 'OpSet2',
       'SensorMeasure20', 'SensorMeasure21', 'OpSet1', 'SensorMeasure6',
       'SensorMeasure10', 'SensorMeasure1', 'SensorMeasure5',
       'SensorMeasure18','RUL']]
```

```python
new_df
```

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn import metrics
import matplotlib.pyplot as plt


all_feature_cols = new_df.columns[:-1]   
target_col = new_df.columns[-1]          

number_of_features = []
train_rmse = []
test_rmse = []

max_features = len(all_feature_cols)     

for i in range(1, max_features + 1):

    number_of_features.append(i)

    
    selected_features = all_feature_cols[:i]

    X = new_df[selected_features]
    y = new_df[target_col]

   
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, shuffle=False
    )

    
    model = RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )
    model.fit(X_train, y_train)

    
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    
    rmse_train_val = np.sqrt(metrics.mean_squared_error(y_train, y_train_pred))
    rmse_test_val = np.sqrt(metrics.mean_squared_error(y_test, y_test_pred))

    train_rmse.append(rmse_train_val)
    test_rmse.append(rmse_test_val)

    print(f"Features Used: {i}")
    print("Selected Columns:", list(selected_features))
    print(f"Train RMSE: {rmse_train_val:.4f}")
    print(f"Test RMSE: {rmse_test_val:.4f}\n")


plt.figure(figsize=(10,7))
plt.plot(number_of_features, train_rmse, marker='o', label='Train RMSE')
plt.plot(number_of_features, test_rmse, marker='o', label='Test RMSE')
plt.xlabel("Number of Features")
plt.ylabel("RMSE")
plt.title("RMSE vs Number of Features (Random Forest)")
plt.legend()
plt.grid(True)
plt.show()
```
